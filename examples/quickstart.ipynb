# PromptPotter Optimizer - Quick Start

This notebook demonstrates basic usage of the PromptPotter Optimizer API.

## Prerequisites
- PromptPotter API running (locally or hosted)
- Python requests library

```python
# Install required packages
!pip install requests pandas -q
```

```python
import requests
import json
from pprint import pprint

# Configure API endpoint
# For local development:
API_URL = "http://localhost:8000/api/v1"

# For hosted service:
# API_URL = "https://your-api.com/api/v1"

# For local with ngrok:
# API_URL = "https://your-ngrok-url.ngrok.io/api/v1"
```

## Step 1: Test API Connection

```python
# Check API health
response = requests.get(f"{API_URL}/health")
print("API Status:")
pprint(response.json())
```

## Step 2: Prepare Your Dataset

Create a dataset with examples. Each example should have:
- Input fields (e.g., "text")
- Expected output (e.g., "expected")

```python
# Example: Sentiment classification dataset
dataset = [
    {"text": "I absolutely love this product!", "expected": "positive"},
    {"text": "This is the worst experience ever", "expected": "negative"},
    {"text": "Amazing quality, highly recommend", "expected": "positive"},
    {"text": "Terrible service and poor quality", "expected": "negative"},
    {"text": "Best purchase I've made this year", "expected": "positive"},
    {"text": "Disappointed and frustrated", "expected": "negative"},
    {"text": "Exceeded all my expectations!", "expected": "positive"},
    {"text": "Would not recommend to anyone", "expected": "negative"},
]

print(f"Dataset size: {len(dataset)} examples")
```

## Step 3: Define Initial Prompt

```python
initial_prompt = "Classify the sentiment of the following text:"
```

## Step 4: Run Optimization

```python
# Optimization request
optimization_request = {
    "initial_prompt": initial_prompt,
    "dataset": dataset,
    "target_metric": "accuracy",
    "model": "gpt-4",
    "max_iterations": 5
}

print("Starting optimization...")
response = requests.post(
    f"{API_URL}/optimize",
    json=optimization_request
)

if response.status_code == 200:
    result = response.json()
    print("\n✓ Optimization completed successfully!")
else:
    print(f"✗ Error: {response.status_code}")
    print(response.text)
```

## Step 5: Review Results

```python
if response.status_code == 200:
    print("\n" + "="*60)
    print("OPTIMIZATION RESULTS")
    print("="*60)
    
    print(f"\nInitial Prompt:\n{initial_prompt}")
    print(f"\nOptimized Prompt:\n{result['optimized_prompt']}")
    
    print(f"\n{'─'*60}")
    print(f"Initial Score:     {result['initial_score']:.2%}")
    print(f"Final Score:       {result['final_score']:.2%}")
    print(f"Improvement:       {result['improvement']:.2f}%")
    print(f"Iterations:        {result['total_iterations']}")
    print(f"Optimization Time: {result['optimization_time']:.2f}s")
    print(f"Model Used:        {result['model_used']}")
```

## Step 6: Analyze Iteration History

```python
if response.status_code == 200:
    print("\nIteration History:")
    print("="*60)
    
    for iteration in result['iterations']:
        print(f"\nIteration {iteration['iteration']}:")
        print(f"  Score: {iteration['score']:.2%}")
        print(f"  Prompt: {iteration['prompt'][:80]}...")
```
